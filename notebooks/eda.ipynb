# Exploratory Data Analysis (EDA) - Used Car Price Prediction

This notebook performs comprehensive Exploratory Data Analysis on the Kaggle used car dataset.

### Objectives:
1. Inspect data shape, types, missing values, and summary statistics.
2. Analyze target variable (`price`) distribution.
3. Investigate relationship between Price and Car Age / Mileage.
4. Analyze Price breakdown by Brand, Fuel Type, Transmission, and Accident History.
5. Examine feature correlations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re

# Set style
sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Load and Inspect Raw Data

```python
df = pd.read_csv('../data/used_cars.csv')
print('Dataset Shape:', df.shape)
df.head()
```

## 2. Missing Values & Data Cleaning

```python
print('--- Missing Values ---')
print(df.isnull().sum())

# Clean Price and Mileage
df['price_num'] = df['price'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False).astype(float)
df['milage_num'] = df['milage'].astype(str).str.replace(' mi.', '', regex=False).str.replace(',', '', regex=False).astype(float)
df['car_age'] = 2024 - df['model_year']
df['milage_per_year'] = df['milage_num'] / (df['car_age'] + 1.0)

df[['price_num', 'milage_num', 'car_age', 'milage_per_year']].describe()
```

## 3. Price Distribution Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(df['price_num'], kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Used Car Price Distribution (Original Scale)')
axes[0].set_xlabel('Price ($)')

sns.histplot(np.log1p(df['price_num']), kde=True, ax=axes[1], color='salmon')
axes[1].set_title('Used Car Price Distribution (Log Scale)')
axes[1].set_xlabel('Log(Price + 1)')

plt.tight_layout()
plt.show()
```

## 4. Price vs Car Age and Mileage

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.scatterplot(data=df, x='car_age', y='price_num', alpha=0.7, ax=axes[0], color='purple')
axes[0].set_title('Price vs Car Age')
axes[0].set_xlabel('Car Age (Years)')
axes[0].set_ylabel('Price ($)')

sns.scatterplot(data=df, x='milage_num', y='price_num', alpha=0.7, ax=axes[1], color='green')
axes[1].set_title('Price vs Mileage')
axes[1].set_xlabel('Mileage (miles)')
axes[1].set_ylabel('Price ($)')

plt.tight_layout()
plt.show()
```

## 5. Price Distribution by Brand & Fuel Type

```python
plt.figure(figsize=(14, 6))
top_brands = df['brand'].value_counts().head(10).index
sns.boxplot(data=df[df['brand'].isin(top_brands)], x='brand', y='price_num', palette='Set2')
plt.title('Price Distribution for Top 10 Car Brands')
plt.xticks(rotation=45)
plt.yscale('log')
plt.ylabel('Price ($) - Log Scale')
plt.show()
```

## 6. Correlation Matrix of Numerical Features

```python
num_cols = ['price_num', 'milage_num', 'model_year', 'car_age', 'milage_per_year']
corr = df[num_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Numerical Feature Correlation Heatmap')
plt.show()
```
